```python
import pandas as pd
import ast
import os
import random
import M1  # Utilizing your M1.py logic for string matching and ultra-metric
import datetime

# 1. Configuration
OUT_DIR = "../data/meal_output_data"
if not os.path.exists(OUT_DIR):
    os.makedirs(OUT_DIR)

# Threshold for similarity between requested category and meal category
string_matching_threshold = 0.5 

# 2. Data Import
# user_meal_requirements contains: user_id, meal_occasion, required_categories
# meal_categories contains: meal_name, categories
REQ_PATH = "../data/input_data/user_meal_requirements.csv"
MEAL_PATH = "../data/input_data/meal_categories.csv"

req_df = pd.read_csv(REQ_PATH)
meal_df = pd.read_csv(MEAL_PATH)

# Preprocess category strings into sets/lists
req_df['required_categories'] = req_df['required_categories'].apply(lambda x: ast.literal_eval(x))
meal_df['categories'] = meal_df['categories'].apply(lambda x: ast.literal_eval(x))

# Map meals to their categories for the ranking algorithm
meal_categories_dict = dict(zip(meal_df['meal_name'], meal_df['categories']))
all_meal_names = meal_df['meal_name'].tolist()

def generate_meal_m1_precise():
    results = []
    print("Start time:\t", datetime.datetime.now())
    
    for _, row in req_df.iterrows():
        user_id = row['user_id']
        occasion = row['meal_occasion']
        required_cats = list(row['required_categories'])
        
        # Step 1: Generate Ranking AND Pseudo-Categories
        # This identifies which meals have categories that string-match the user's requirements
        ranking, pseudo_meal_categories = M1.string_matching_ranking(
            all_researcher_skills=meal_categories_dict, # Mapping meal->categories
            proposal_skills=required_cats,               # The user's requested categories
            pseudo_researcher_skills={}, 
            matching_threshold=string_matching_threshold
        )
        
        # Step 2: Pick a 'target meal' to lead the bundle (or pick randomly from pool)
        target_meal = random.choice(all_meal_names)
        
        # Step 3: Create a bundle (team) of 3 meals using the Ranking
        # M1 logic samples additional items from the top-ranked matches
        bundles_list = M1.create_teams_for_each_person(
            ranking=ranking,
            target_researcher=target_meal,
            num_of_teams=1,
            defaultSizeFlag=True,
            defaultSize=3  # Standardizing on a 3-meal bundle
        )
        
        if not bundles_list:
            continue
            
        meal_bundle = bundles_list[0]
        
        # Step 4: Apply Ultra-Metric using Pseudo-Categories
        # Calculates goodness based only on the categories that matched the user's request
        goodness = M1.apply_ultra_metric(
            set(required_cats), 
            meal_bundle, 
            pseudo_meal_categories
        )
        
        results.append({
            'user_id': user_id,
            'meal_occasion': occasion,
            'lead_meal': target_meal,
            'meal_bundle': meal_bundle,
            'goodness_score': goodness
        })
        
    print("End time:\t", datetime.datetime.now())
    return pd.DataFrame(results)

# 3. Execution
final_results = generate_meal_m1_precise()
output_path = os.path.join(OUT_DIR, 'meal_results_m1_precise.csv')
final_results.to_csv(output_path, index=False)

print(f"Meal recommendation complete. {len(final_results)} bundles saved to {output_path}")
```
