```python
import pandas as pd
import numpy as np
import ast
import os
import random
import M1  # Utilizing your M1.py for the Ultra-Metric
import datetime

# 1. Configuration
OUT_DIR = "../data/meal_output_data"
if not os.path.exists(OUT_DIR):
    os.makedirs(OUT_DIR)

# Knapsack Parameters
CAPACITY_LIMIT = 5    # Max meals per bundle
COST_MULTIPLIER = 0.1 # Higher = smaller, more efficient bundles
STOCHASTIC = True     # Adds slight randomness to selection for variety

# 2. Data Import
REQ_PATH = "../data/input_data/user_meal_requirements.csv"
MEAL_PATH = "../data/input_data/meal_categories.csv"

req_df = pd.read_csv(REQ_PATH)
meal_df = pd.read_csv(MEAL_PATH)

def parse_set(x):
    if pd.isna(x): return set()
    try:
        val = ast.literal_eval(x)
        return set(val) if isinstance(val, (list, set)) else set()
    except:
        return set()

req_df['req_set'] = req_df['required_categories'].apply(parse_set)
meal_df['cat_set'] = meal_df['categories'].apply(parse_set)

meal_to_cats = dict(zip(meal_df['meal_name'], meal_df['cat_set']))
all_meal_names = list(meal_to_cats.keys())

# 3. Helper: Pick the best lead meal based on overlap
def pick_best_lead_meal(required_set):
    best_meal = None
    max_overlap = -1
    for meal, cats in meal_to_cats.items():
        overlap = len(cats & required_set)
        if overlap > max_overlap:
            max_overlap = overlap
            best_meal = meal
    return best_meal if best_meal else random.choice(all_meal_names)

# 4. Knapsack Solver for Meals
def solve_meal_knapsack(target_meal, required_set, capacity=5, cost_mult=0.1, stochastic=True):
    bundle = [target_meal]
    current_covered = meal_to_cats[target_meal] & required_set
    
    while len(bundle) < capacity:
        candidates = []
        for meal, cats in meal_to_cats.items():
            if meal in bundle: continue
            
            # Marginal gain: count of NEW categories covered
            new_cats = (cats & required_set) - current_covered
            gain = len(new_cats)
            
            # Net gain = Gain - (Penalty for bundle size)
            net_gain = gain - (cost_mult * len(bundle))
            
            if net_gain > 0:
                candidates.append((meal, net_gain, new_cats))
        
        if not candidates:
            break
            
        # Sort by net gain
        candidates.sort(key=lambda x: x[1], reverse=True)
        
        if stochastic:
            # Pick from the top 3 best fits to ensure recommendation variety
            selection = random.choice(candidates[:min(3, len(candidates))])
            chosen_meal, _, new_cats_added = selection
        else:
            chosen_meal, _, new_cats_added = candidates[0]
            
        bundle.append(chosen_meal)
        current_covered.update(new_cats_added)
        
    return bundle

# 5. Execution Loop
def run_meal_knapsack_teaming():
    results = []
    print("Start time:\t", datetime.datetime.now())
    
    for idx, row in req_df.iterrows():
        user_id = row['user_id']
        occasion = row['meal_occasion']
        req_set = row['req_set']
        
        # Step A: Find the best starting meal
        lead = pick_best_lead_meal(req_set)
        
        # Step B: Build the optimized bundle
        bundle = solve_meal_knapsack(
            target_meal=lead, 
            required_set=req_set, 
            capacity=CAPACITY_LIMIT, 
            cost_mult=COST_MULTIPLIER,
            stochastic=STOCHASTIC
        )
        
        # Step C: Evaluate with Ultra-Metric
        goodness = M1.apply_ultra_metric(req_set, bundle, meal_to_cats)
        
        results.append({
            'user_id': user_id,
            'meal_occasion': occasion,
            'lead_researcher': lead, # keeping column names consistent with your previous request
            'team': bundle,           # keeping column names consistent
            'goodness_score': goodness
        })
        
        if (idx + 1) % 50 == 0:
            print(f"Processed {idx+1}/{len(req_df)} users...")

    return pd.DataFrame(results)

# 6. Save Results
final_results = run_meal_knapsack_teaming()
output_file = os.path.join(OUT_DIR, 'meal_results_knapsack.csv')
final_results.to_csv(output_file, index=False)

print(f"\nKnapsack Meal Recommendation complete. Saved to {output_file}")
print("Average Goodness:", round(final_results['goodness_score'].mean(), 4))
```
