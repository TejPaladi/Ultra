```python
import pandas as pd
import ast
import os
import random
import M3  # Your provided M3.py file
import datetime
import nlp_techniques # Required for N-gram generation

# 1. Configuration
OUT_DIR = "../data/meal_output_data"
if not os.path.exists(OUT_DIR):
    os.makedirs(OUT_DIR, exist_ok=True)

# M3 specific threshold for meal category similarity
string_matching_threshold = 0.3 

# 2. Data Import
REQ_PATH = "../data/input_data/user_meal_requirements.csv"
MEAL_PATH = "../data/input_data/meal_categories.csv"

req_df = pd.read_csv(REQ_PATH)
meal_df = pd.read_csv(MEAL_PATH)

# 3. Helper for Meal Name Normalization (Matching the Bandit DB format)
def normalize_meal_name(name_str):
    return name_str.strip().replace(',', '')\
        .replace('.', '')\
        .replace('(', '')\
        .replace(')', '')\
        .replace(' ', '_')\
        .replace('-', '_')\
        .replace('\'', '_')\
        .lower()

# 4. M3 Category Expansion (N-grams) & Preprocessing
# Treating meal categories as "researcher skills"
m3_meal_categories = {}
for i in range(len(meal_df)):
    meal_name = meal_df.iloc[i]['meal_name']
    cats_data = meal_df.iloc[i]['categories']
    
    # Load categories as set/list
    categories = ast.literal_eval(cats_data)
    
    # Generate bigrams (2-grams) for better semantic matching
    n_gram_interests = []
    for cat in categories:
        n_gram_interests.append(nlp_techniques.generate_N_grams(cat, ngram=2))
    
    # Flatten list of lists
    n_gram_interests = [item for sublist in n_gram_interests for item in sublist]
    
    # Combine original categories with expanded N-grams
    m3_meal_categories[meal_name] = set(categories).union(n_gram_interests)

# 5. Parse Bandit Results (results_team.db for meals)
# Assuming the bandit model was trained using meal/user IDs
pos_bundles = {}
neg_bundles = {}
RESULTS_PATH = "./boosted_results/test/results_meal_team.db" # Adjusted path

if os.path.exists(RESULTS_PATH):
    with open(RESULTS_PATH, 'r') as file:
        lines = [line.rstrip() for line in file]
        for line in lines:
            if not line: continue
            
            if line.startswith("team"):
                posFlag = True
                prefix_len = 5 # 'team('
            elif line.startswith("!team"):
                posFlag = False
                prefix_len = 6 # '!team('
            else:
                continue
                
            tokens = line.split(" ")
            if len(tokens) < 3: continue
            
            # Format: team(requirement_id, meal_norm) likelihood
            req_id = tokens[0][prefix_len:-1]
            meal_norm = tokens[1][:-1]
            likelihood = float(tokens[2])
            
            target_dict = pos_bundles if posFlag else neg_bundles
            if req_id not in target_dict:
                target_dict[req_id] = {}
            target_dict[req_id][meal_norm] = likelihood

# Mapping for bandit lookups
norm_to_real_meal = {normalize_meal_name(m): m for m in m3_meal_categories.keys()}
all_meal_names = list(m3_meal_categories.keys())

def generate_meal_m3_boosted(pos_bundles, neg_bundles):
    results = []
    print("Start time:\t", datetime.datetime.now())

    for _, row in req_df.iterrows():
        user_id = str(row['user_id'])
        occasion = row['meal_occasion']
        req_id = user_id # Using user_id as the ID for bandit matching
        
        req_cats = list(ast.literal_eval(row['required_categories']))

        # Step 1: String Matching & Pseudo-Category Filtering
        ranking, pseudo_categories = M3.string_matching_ranking(
            m3_meal_categories, req_cats, {}, matching_threshold=string_matching_threshold
        )

        # Step 2: Boosted Bandit Selection (Prioritize "Positive" meal recommendations)
        meal_bundle = []
        if req_id in pos_bundles:
            pos_meals_norm = list(pos_bundles[req_id].keys())
            pos_meals = [norm_to_real_meal[m] for m in pos_meals_norm if m in norm_to_real_meal]
            if pos_meals:
                # Create a bundle of up to 3 meals from the positive list
                meal_bundle = random.sample(pos_meals, min(len(pos_meals), 3))
        
        # Fallback to String Matching Ranking and Negative Filtering
        if not meal_bundle:
            lead_meal = random.choice(all_meal_names)
            # Create a bundle using M3 ranking logic
            m3_bundles = M3.create_teams_for_each_person(ranking, lead_meal, 1, defaultSizeFlag=True, defaultSize=3)
            if m3_bundles:
                meal_bundle = m3_bundles[0]
                # Filter out meals the bandit model flagged as negative
                if req_id in neg_bundles:
                    neg_meals_norm = list(neg_bundles[req_id].keys())
                    meal_bundle = [m for m in meal_bundle if normalize_meal_name(m) not in neg_meals_norm]

        # Step 4: Final Scoring with Ultra-Metric
        if meal_bundle:
            goodness = M3.apply_ultra_metric(set(req_cats), meal_bundle, pseudo_categories)
            results.append({
                'user_id': user_id,
                'meal_occasion': occasion,
                'lead_meal': meal_bundle[0],
                'meal_bundle': meal_bundle, 
                'goodness_score': goodness
            })

    print("End time:\t", datetime.datetime.now())
    return pd.DataFrame(results)

# 6. Execution and Saving to CSV
final_results_df = generate_meal_m3_boosted(pos_bundles, neg_bundles)
save_path = os.path.join(OUT_DIR, 'meal_results_m3_boosted.csv')
final_results_df.to_csv(save_path, index=False)

print(f"M3 Meal Recommendation complete. {len(final_results_df)} bundles saved to {save_path}")
```
