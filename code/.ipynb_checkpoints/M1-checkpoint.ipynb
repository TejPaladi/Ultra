```python
import pandas as pd
import ast
import os
import random
import M1  # Utilizing your provided M1.py file
import datetime

# Use the threshold from your original snippet
string_matching_threshold = 0.5 

# 1. Configuration and Directory Setup
# Ensure the output directory exists
OUT_DIR = "../data/output_data"
if not os.path.exists(OUT_DIR):
    os.makedirs(OUT_DIR)

RESEARCHER_SKILLS_PATH = "../data/input_data/Set_4/large_researcher_skills.csv"
PROPOSAL_SKILLS_PATH   = "../data/input_data/Set_4/large_proposal_skills.csv"

# 2. Data Import 
# Load your uploaded data files
proposals_df = pd.read_csv(PROPOSAL_SKILLS_PATH) 
researchers_df = pd.read_csv(RESEARCHER_SKILLS_PATH) 

# Preprocess skills from string to actual sets/lists
proposals_df['skills'] = proposals_df['skills'].apply(lambda x: ast.literal_eval(x))
researchers_df['skills'] = researchers_df['skills'].apply(lambda x: ast.literal_eval(x))

# Create dictionaries for M1 processing
researcher_skills_dict = dict(zip(researchers_df['researcher_name'], researchers_df['skills']))
all_names = researchers_df['researcher_name'].tolist()

def generate_m1_teams_final():
    results = []
    
    print("Start time:\t", datetime.datetime.now()) #
    
    # Loop through each proposal
    for _, prop in proposals_df.iterrows():
        prop_link = prop['nsf_proposal_links_v0']
        prop_skills = list(prop['skills'])
        
        # Step 1: Rank researchers based on string matching (0.5 threshold)
        # We call this once per proposal to get the ranking for ALL researchers
        ranking, _ = M1.string_matching_ranking(
            all_researcher_skills=researcher_skills_dict,
            proposal_skills=prop_skills,
            pseudo_researcher_skills={}, # Passing empty dict triggers calculation
            matching_threshold=string_matching_threshold
        )
        
        # Step 2: Pick one random 'target researcher' to lead the team
        # To ensure we get exactly 1 team per proposal (500 rows total)
        lead_researcher = random.choice(all_names)
        
        # Step 3: Create 1 team using the M1 ranking
        # This will sample from the top 20 ranked researchers for this proposal
        teams_list = M1.create_teams_for_each_person(
            ranking=ranking,
            target_researcher=lead_researcher,
            num_of_teams=1
        )
        
        if not teams_list:
            continue
            
        team = teams_list[0]
        
        # Step 4: Calculate the goodness score using the Ultra-Metric
        # Evaluating the chosen team against the original proposal skills
        goodness = M1.apply_ultra_metric(set(prop_skills), team, researcher_skills_dict)
        
        results.append({
            'nsf_proposal_links_v0': prop_link,
            'lead_researcher': lead_researcher,
            'team': team,
            'goodness_score': goodness
        })
        
    print("End time:\t", datetime.datetime.now()) #
    return pd.DataFrame(results)

# 3. Execution
final_results = generate_m1_teams_final()

# Save output to your requested directory
output_path = os.path.join(OUT_DIR, 'teaming_results_m1.csv')
final_results.to_csv(output_path, index=False)

print(f"Process complete. {len(final_results)} rows saved to {output_path}")
```
